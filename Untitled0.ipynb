```python
from pathlib import Path

parquet_files = sorted(Path("/content/data").rglob("*.parquet"))

print(f"Found {len(parquet_files)} Parquet files")
for path in parquet_files[:20]:
    print(path)
```

```python
import pandas as pd
import numpy as np
import pyarrow.parquet as pq
from collections import Counter

SEED = 42
BATCH_SIZE = 100_000
MAX_PER_FILE_CATEGORY = 5_000
SAMPLE_PER_BATCH = 1_000

def categorize_label(label):
    text = str(label).strip().lower()

    if "benign" in text or "normal" in text:
        return "Normal"
    if "ddos" in text:
        return "DDoS"
    if "dos" in text:
        return "DoS"
    return "Other attack"

label_counts = Counter()
sample_pools = {}

for file_path in parquet_files:
    print("Reading:", file_path.name)
    parquet_file = pq.ParquetFile(file_path)

    for batch in parquet_file.iter_batches(batch_size=BATCH_SIZE):
        chunk = batch.to_pandas()
        chunk.columns = [str(c).strip() for c in chunk.columns]
        chunk = chunk.loc[:, ~chunk.columns.duplicated()]

        label_col = next(
            (c for c in chunk.columns if c.lower() == "label"),
            None
        )

        if label_col is None:
            raise ValueError(
                f"No Label column found in {file_path.name}. "
                f"First columns: {chunk.columns.tolist()[:20]}"
            )

        chunk["raw_label"] = chunk[label_col].astype(str).str.strip()
        chunk["category"] = chunk["raw_label"].map(categorize_label)
        chunk["source_file"] = file_path.name

        # Count labels across all rows, not just the sample.
        label_counts.update(chunk["raw_label"].value_counts().to_dict())

        # Keep a sample of the categories relevant to this analysis.
        for category in ["Normal", "DoS", "DDoS"]:
            group = chunk.loc[chunk["category"] == category].copy()

            if group.empty:
                continue

            if len(group) > SAMPLE_PER_BATCH:
                group = group.sample(
                    n=SAMPLE_PER_BATCH,
                    random_state=SEED
                )

            key = (file_path.name, category)
            existing = sample_pools.get(key)

            pool = group if existing is None else pd.concat(
                [existing, group],
                ignore_index=True
            )

            if len(pool) > MAX_PER_FILE_CATEGORY:
                pool = pool.sample(
                    n=MAX_PER_FILE_CATEGORY,
                    random_state=SEED
                )

            sample_pools[key] = pool

if not sample_pools:
    raise ValueError(
        "No Normal, DoS, or DDoS rows were sampled. "
        "Check the label column and label values."
    )

df = pd.concat(sample_pools.values(), ignore_index=True)

print("Sample shape:", df.shape)
display(df.head())
```

```python
display(
    pd.Series(label_counts)
      .sort_values(ascending=False)
      .head(30)
      .to_frame("row_count")
)

display(df["category"].value_counts().to_frame("sample_rows"))
```

```python
metadata_cols = {"raw_label", "category", "source_file"}

for col in df.columns:
    if col not in metadata_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

df[numeric_cols] = df[numeric_cols].replace(
    [np.inf, -np.inf],
    np.nan
)

# Drop columns with no usable numeric values.
numeric_cols = [col for col in numeric_cols if not df[col].isna().all()]

print("Rows in sample:", len(df))
print("Numeric features:", len(numeric_cols))
print("Numeric feature examples:", numeric_cols[:30])
```

```python
import re

def normalize_name(text):
    return re.sub(r"[^a-z0-9]", "", str(text).lower())

available = {normalize_name(col): col for col in numeric_cols}

expected_features = [
    "Flow Byts/s",
    "Flow Pkts/s",
    "Fwd Pkts/s",
    "Bwd Pkts/s",
    "Tot Fwd Pkts",
    "Tot Bwd Pkts",
    "TotLen Fwd Pkts",
    "TotLen Bwd Pkts",
]

traffic_features = [
    available[normalize_name(name)]
    for name in expected_features
    if normalize_name(name) in available
]

print("Traffic-volume features found:", traffic_features)
```

```python
if traffic_features:
    display(
        df.groupby("category")[traffic_features]
          .median()
          .T
          .rename_axis("feature")
    )
else:
    print("No expected traffic-volume features were found.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for feature in traffic_features:
    plot_df = df[["category", feature]].dropna().copy()
    plot_df[feature] = np.log1p(plot_df[feature].clip(lower=0))

    plt.figure(figsize=(9, 4))
    sns.histplot(
        data=plot_df,
        x=feature,
        hue="category",
        stat="density",
        common_norm=False,
        element="step",
        bins=50
    )
    plt.title(f"{feature} by category — log scale")
    plt.xlabel(f"log(1 + {feature})")
    plt.tight_layout()
    plt.show()
```

```python
attack_df = df[df["category"].isin(["DoS", "DDoS"])].copy()
normal_df = df[df["category"] == "Normal"].copy()

if normal_df.empty or attack_df.empty:
    print("The sample needs both Normal and DoS/DDoS rows for this comparison.")
else:
    normal_median = normal_df[numeric_cols].median()
    attack_median = attack_df[numeric_cols].median()

    normal_iqr = (
        normal_df[numeric_cols].quantile(0.75)
        - normal_df[numeric_cols].quantile(0.25)
    )
    attack_iqr = (
        attack_df[numeric_cols].quantile(0.75)
        - attack_df[numeric_cols].quantile(0.25)
    )

    scale = (normal_iqr + attack_iqr) / 2

    feature_comparison = pd.DataFrame({
        "normal_median": normal_median,
        "attack_median": attack_median,
        "scaled_median_difference": (
            (attack_median - normal_median).abs() / scale.replace(0, np.nan)
        )
    })

    display(
        feature_comparison
          .sort_values("scaled_median_difference", ascending=False)
          .head(20)
    )
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

model_df = df[df["category"].isin(["Normal", "DoS", "DDoS"])].copy()

y = (model_df["category"] != "Normal").astype(int)
X = model_df[numeric_cols].copy()

# Remove timestamp-like numeric columns, which can give an unreliable baseline.
drop_cols = [
    col for col in X.columns
    if any(word in col.lower() for word in ["timestamp", "date", "time"])
]
X = X.drop(columns=drop_cols, errors="ignore")

# Replace infinite values, remove mostly empty features, and fill remaining gaps.
X = X.replace([np.inf, -np.inf], np.nan)
X = X.loc[:, X.isna().mean() < 0.50]
X = X.fillna(X.median()).fillna(0)

# Remove constant features.
X = X.loc[:, X.nunique() > 1]

print("Rows:", len(X))
print("Class counts (0=Normal, 1=DoS/DDoS):")
print(y.value_counts())

if y.nunique() < 2:
    raise ValueError("The sample must contain both Normal and attack rows.")

# Limit training size to keep Colab responsive.
MAX_MODEL_ROWS = 100_000
if len(X) > MAX_MODEL_ROWS:
    X, _, y, _ = train_test_split(
        X,
        y,
        train_size=MAX_MODEL_ROWS,
        stratify=y,
        random_state=SEED
    )

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    stratify=y,
    random_state=SEED
)

model = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced_subsample",
    min_samples_leaf=2,
    random_state=SEED,
    n_jobs=-1
)

model.fit(X_train, y_train)
predictions = model.predict(X_test)

print(classification_report(
    y_test,
    predictions,
    target_names=["Normal", "DoS/DDoS"],
    zero_division=0
))

print("Confusion matrix:")
print(confusion_matrix(y_test, predictions))
```

```python
importances = pd.Series(
    model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

display(importances.head(20).to_frame("importance"))

importances.head(20).sort_values().plot(
    kind="barh",
    figsize=(8, 6),
    title="Most important Random Forest features"
)
plt.tight_layout()
plt.show()
```

```python
from sklearn.ensemble import IsolationForest

# Use the same cleaned feature columns as the classifier.
X_anomaly = model_df[X.columns].copy()
X_anomaly = X_anomaly.replace([np.inf, -np.inf], np.nan)
X_anomaly = X_anomaly.fillna(X.median()).fillna(0)

y_anomaly = (model_df["category"] != "Normal").astype(int)

X_train_a, X_test_a, y_train_a, y_test_a = train_test_split(
    X_anomaly,
    y_anomaly,
    test_size=0.25,
    stratify=y_anomaly,
    random_state=SEED
)

normal_training_rows = X_train_a.loc[y_train_a == 0]

# Limit the normal training set for speed.
if len(normal_training_rows) > 10_000:
    normal_training_rows = normal_training_rows.sample(
        n=10_000,
        random_state=SEED
    )

iso = IsolationForest(
    n_estimators=150,
    contamination="auto",
    random_state=SEED,
    n_jobs=-1
)

iso.fit(normal_training_rows)

# Isolation Forest marks outliers as -1. Convert those to attack predictions.
anomaly_predictions = (iso.predict(X_test_a) == -1).astype(int)

print(classification_report(
    y_test_a,
    anomaly_predictions,
    target_names=["Normal", "DoS/DDoS"],
    zero_division=0
))

print("Confusion matrix:")
print(confusion_matrix(y_test_a, anomaly_predictions))
```
